```python
num_cols1 = ['cylinders','acceleration','origin','model year']                         #this list is for symmetric numeric columns
log_cols1 = ['mpg','horsepower','displacement','weight']   #this list is for skewed numeric columns
cat_cols1 = ['car name']     #this list for the class columns

target1 = 'Documents/MASTER 2024/Summer 2024/Practical Data Analytics/auto-mpg.csv'
```

```python
#processing pipeline
from sklearn.preprocessing import OneHotEncoder, FunctionTransformer
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
import numpy as np

num_pipeline = Pipeline([                           #now we need a small pipeline for numeric columns since it has two steps
    ('impute', SimpleImputer(strategy='median')),   #this step will impute missing values using column medians
    ('standardize', StandardScaler())               #this step will scale all numeric columns
])

def log_transformation(x):
    return np.log(x+0.1)

log_pipeline = Pipeline([                                        #now we need a small pipeline for numeric columns since it has two steps
    ('impute', SimpleImputer(strategy='median')),                #this step will impute missing values using column medians
    ('log transform', FunctionTransformer(log_transformation)), #this include a log transformation into the pipeline
    ('standardize', StandardScaler())                            #this step will scale all numeric columns
])

processing_pipeline = ColumnTransformer([                                                      #this transformer merges the processed numeric columns and class columns
    ('numeric', num_pipeline, num_cols1),                                                       #numeric columns
    ('log', log_pipeline, log_cols1),                                                           #log columns
    ('class', OneHotEncoder(max_categories=5, handle_unknown='infrequent_if_exist'), cat_cols1) #encoder to transform class columns to numeric, this will automatically handle missing data
  ])
```

```python
processing_pipeline
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

# Define the target column name
target_column = 'mpg'

logistic_pipeline = Pipeline([
    ('processing', processing_pipeline),
    ('modeling', LinearRegression())  # Use LinearRegression for regression problems
])

# Fit the pipeline
logistic_pipeline.fit(traindata1, traindata1[target_column])
```

```python
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Define the target column name
target_column = 'mpg'

# Binarize the target column
median_mpg = traindata1[target_column].median()
traindata1['mpg_category'] = np.where(traindata1[target_column] > median_mpg, 1, 0)

# Define the logistic regression pipeline
logistic_pipeline = Pipeline([
    ('processing', processing_pipeline),
    ('modeling', LogisticRegression())
])

# Fit the pipeline
logistic_pipeline.fit(traindata1, traindata1['mpg_category'])

# Make predictions
logistic_pred = logistic_pipeline.predict(traindata1)

# Calculate accuracy score
accuracy = accuracy_score(traindata1['mpg_category'], logistic_pred)
print(f"Accuracy Score: {accuracy}")
```

```python
from sklearn.tree import DecisionTreeClassifier

# Define the target column name
target_column = 'mpg'

# Ensure the target column is binarized as before
traindata1['mpg_category'] = np.where(traindata1[target_column] > median_mpg, 1, 0)

tree_pipeline = Pipeline([
    ('processing', processing_pipeline),
    ('modeling', DecisionTreeClassifier())
])

tree_pipeline.fit(traindata1, traindata1['mpg_category'])
```

```python
# Make predictions using the tree pipeline
tree_pred = tree_pipeline.predict(traindata1)

# Calculate accuracy score
tree_accuracy = accuracy_score(traindata1['mpg_category'], tree_pred)
print(f"Tree Accuracy Score: {tree_accuracy}")
```

```python
from sklearn.model_selection import train_test_split

# Load and prepare data
df = pd.read_csv('Documents/MASTER 2024/Summer 2024/Practical Data Analytics/auto-mpg.csv')
median_mpg = df['mpg'].median()
df['mpg_category'] = np.where(df['mpg'] > median_mpg, 1, 0)

# Split the data into training and testing sets
traindata1, testdata1 = train_test_split(df, test_size=0.2, random_state=42)
```
